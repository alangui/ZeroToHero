```python
import sentencepiece as spm

sp = spm.SentencePieceProcessor(model_file='zh_bpe.model')
print(sp.vocab_size())            # 应该输出 16000
print(sp.piece_to_id('[BR]'))     # 应该输出 3
```

```python
with open('../wiki_corpus_0.1b_clean.txt', 'r', encoding='utf-8') as f:
    text = f.read()

data = sp.encode(text)
chars = len(text)
tokens = len(data)
print(f"字符数: {chars}")
print(f"BPE token 数: {tokens}")
print(f"压缩率: {tokens / chars:.2f}")
```
