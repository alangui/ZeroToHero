```python
import sentencepiece as spm

sp = spm.SentencePieceProcessor(model_file='zh_bpe.model')
print(sp.vocab_size())            # 应该输出 16000
print(sp.piece_to_id('[BR]'))     # 应该输出 3
```

```python
# 看词表大小
print(sp.vocab_size())  # 你日志显示 16000，这个没问题

# 测试一个常见中文词被切成几段
test = "中华人民共和国"
pieces = sp.encode_as_pieces(test)
print(f"'{test}' -> {pieces}")  
```

```python
for word in ["中国", "学习", "的", "了", "我们"]:
    ids = sp.encode(word)
    print(f"'{word}' ({len(word)}字) -> {len(ids)} tokens: {sp.encode_as_pieces(word)}")
```

```python
with open('clean_chars.txt', 'r', encoding='utf-8') as f:
    text = f.read()

data = sp.encode(text)
chars = len(text)
tokens = len(data)
print(f"字符数: {chars}")
print(f"BPE token 数: {tokens}")
print(f"压缩率: {tokens / chars:.2f}")
```
